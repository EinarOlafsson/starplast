# Start from a gene list
Four strategies take a list: 02 looks for the cluster that holds it, 20 learns what makes it special from positives alone, 24 describes what it has in common, and 25 grows it along the networks. Here the list is the 135 genes of PM - integral; hold out the compartment column so no strategy can read the list back from the label that defined it.


```python
from starplast import strategies as S
ctx = S.shipped('Tg')
cat, members = S.example_set(ctx, 'compartment')
genes = list(ctx.gene_ids[members])
print(cat, len(genes), genes[:5])
```

```python
hunt = S.run('geneset_hunt', genes=genes, exclude='compartment',
             n_neighbors='15, 50', min_dist='0.0', min_cluster_size='20, 50',
             selection='eom, leaf', sample=2000)
print(hunt.summary)
```

```python
hunt.tables['candidates'].head(10)
```

## Three more ways to use a list


```python
pu = S.run('positive_unlabeled', genes=genes, exclude='compartment')
pu.tables['candidates'].head(8)
```

```python
profile = S.run('set_enrichment', genes=genes, exclude='compartment')
profile.tables['profile'].head(12)
```

```python
grown = S.run('seed_expansion', genes=genes, exclude='compartment')
grown.tables['candidates'].head(8)
```
