# Explore the tables from Python
Everything the window shows is available from Python. A `Context` is one organism's gene table, its measured networks and the leakage guard every analysis applies.


```python
from starplast import strategies as S
ctx = S.shipped('Tg')
print(ctx.organism, ctx.n, 'genes,', len(ctx.nodes.columns), 'columns')
```

```python
hits = ctx.nodes[ctx.nodes['product'].str.contains('microneme protein', na=False)]
hits[['gene_id', 'product', 'compartment', 'fit_invitro_hff']].head(8)
```

```python
ctx.truth('compartment').value_counts().head(10)
```

```python
print(len(ctx.categorical_columns()), 'labels can be held out, e.g.', ctx.categorical_columns()[:6])
print(len(ctx.numeric_columns()), 'measurements are available as features')
```

## What a held-out analysis never sees
Holding a label out removes the label, anything that restates it (measured association of 0.8 or more, on values or ranks), the experiment that produced it, the same quantity measured another way, and any edge layer built from it. This is what makes a recovered label mean something.


```python
sorted(ctx.banned('compartment'))
```

```python
print('edge layers:', ctx.layers())
print('banned when compartment is held out:', sorted(ctx.banned_layers('compartment')))
```
