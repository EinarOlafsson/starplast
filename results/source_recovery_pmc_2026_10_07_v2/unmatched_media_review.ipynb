# Review unmatched named supplements and available primary media


An unmatched historical filename does not show that the paper has no data. Keep every alternative filename visible for source-specific content review; do not rename a different table to pass the exact-name gate.

```python
import json
from pathlib import Path
from urllib.parse import urlparse
root=Path('results/source_recovery_pmc_2026_10_07_v2')
rows=json.loads((root/"source_locations.json").read_text())
for row in rows:
    if row["status"] != "named_file_or_published_version_unresolved": continue
    print(row["source_id"], "wanted", row["filename"])
    for file in root.glob(row["pmcid"]+".*.json"):
        metadata=json.loads(file.read_text())
        print(file.name,metadata["pmid"],metadata["is_manuscript"], [Path(urlparse(u).path).name for u in metadata["media_urls"] if Path(urlparse(u).path).suffix.lower() in {".xls",".xlsx",".csv",".tsv",".zip",".pdf"}])
```
