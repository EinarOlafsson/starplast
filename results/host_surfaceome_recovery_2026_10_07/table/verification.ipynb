# Publisher alias evidence identities and metadata normalization


The constructed HEAD metadata had its trailing JSON newline normalized after initial serialization; no response field or source byte changed. Pin exact raw/cache/code identities and the equivalence result before final ledger admission.

```python
import hashlib,json
from pathlib import Path
from starplast import organisms as O,paths
p=Path("results/host_surfaceome_recovery_2026_10_07/table")
e=json.loads((p/"equivalence.json").read_text())
cache=Path(paths.cache_file(O.HOST_TABLES[O.MOUSE]))
inputs=[Path(e["file"]["path"]),cache,p/"code/host.py",p/"table_review.json",p/"equivalence_code_identity.json"]
manifest={"input_sha256":{str(path):hashlib.sha256(path.read_bytes()).hexdigest() for path in inputs},"outputs":{"equivalence.json":hashlib.sha256((p/"equivalence.json").read_bytes()).hexdigest()},"runtime_numerical_changes":"none"}
(p/"equivalence_manifest.json").write_text(json.dumps(manifest,indent=2)+chr(10))
assert manifest["input_sha256"][e["file"]["path"]]==e["file"]["sha256"]
print({"source_id":e["source_id"],"exact_installed_fields":e["installed_reproduction"],"original_filename":e["published_filename"],"negative_semantics":e["negative_semantics"]})
```
