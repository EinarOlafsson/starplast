# Primary repository metadata for RBC and rhoptry candidate admission


Identifiers are taken from the verified registry/audit and primary publisher/PubMed records. Download repository metadata before considering any assay data. No source promotion follows from a title or article link.

```python
import urllib.request,json,hashlib
from pathlib import Path
output=Path('results/host_candidate_deposits_2026_10_07')
urls = {'rbc_pride': 'https://www.ebi.ac.uk/pride/ws/archive/v2/projects/PXD005200', 'rbc_files': 'https://www.ebi.ac.uk/pride/ws/archive/v2/files/byProject?accession=PXD005200&pageSize=1000', 'rhoptry_pubmed': 'https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi?db=pubmed&id=42791346&retmode=xml'}
retrievals={}
for name,url in urls.items():
    try:
        request=urllib.request.Request(url, headers={'User-Agent':'starplast-dataset-review/1'})
        with urllib.request.urlopen(request, timeout=25) as response:
            data=response.read(8 << 20)
        path=output/(name+('.xml' if name.endswith('pubmed') else '.json'))
        path.write_bytes(data)
        retrievals[name]={'url':url,'bytes':len(data),'sha256':hashlib.sha256(data).hexdigest(),'status':'retrieved'}
    except Exception as exc:
        retrievals[name]={'url':url,'status':'failed','error':str(exc)}
(output/'requests.json').write_text(json.dumps(retrievals, indent=2)+'\n')
retrievals
```
