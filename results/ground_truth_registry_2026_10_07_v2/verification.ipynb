# Ground-truth registry integrity and compatibility verification


Verify the reviewed census and original diagnostic code identities. No fitting or independent biological accuracy claim.

```python
import ast, hashlib, json
from pathlib import Path
from starplast.ground_truth import read_registry
p=Path("results/ground_truth_registry_2026_10_07_v2")
m=json.loads((p/"manifest.json").read_text())
for filename, digest in m["input_sha256"].items():
    assert hashlib.sha256(Path(filename).read_bytes()).hexdigest() == digest
for filename, digest in m["outputs"].items():
    assert hashlib.sha256((p/filename).read_bytes()).hexdigest() == digest
entries,refs=read_registry(p/"registry.json")
for filename in ("starplast/ground_truth.py", "scripts/build_ground_truth_registry.py"):
    ast.parse(Path(filename).read_text(),feature_version=(3,10))
old=Path("results/ground_truth_registry_2026_10_07")
previous=json.loads((old/"manifest.json").read_text())
for filename in ("ground_truth.py", "build_ground_truth_registry.py"):
    archived=old/"code"/filename
    assert hashlib.sha256(archived.read_bytes()).hexdigest() in previous["input_sha256"].values()
print(m["summary"])
print("All frozen inputs, output checksums, cohorts and Python 3.10 syntax verified")
```
