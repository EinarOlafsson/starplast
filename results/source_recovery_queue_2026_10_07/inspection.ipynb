# Remaining source-recovery queue inspection


Inspect unbound registered archive paths and URLs. Installed caches are not raw source bindings. No acquisition or promotion.

```python
import json
from pathlib import Path
from starplast import datasets as D
base=Path("results/source_recovery_2026_10_07")
bindings=json.loads(Path("results/source_recovery_pmc_2026_10_07_v2/bindings.json").read_text())
rows=json.loads((base/"source_locations.json").read_text())
status={r["source_id"]:r["status"] for r in rows}
for d in D.REGISTRY:
    if d.key not in bindings and status[d.key] != "installed_cache":
        print(d.key, status[d.key], d.path, d.url)
print("Bound processed sources:",len(bindings))
```
