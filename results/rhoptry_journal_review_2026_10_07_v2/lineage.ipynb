# Rhoptry version lineage, primary dates and citation snapshots


Journal and preprint are linked versions of one experiment, not independent source competitors. Do not reset publication age in 2026, sum potentially overlapping citation counts, or treat source protein-row coverage as an assayed universe.

```python
from pathlib import Path
import json,urllib.request,urllib.parse,hashlib
from datetime import datetime,timezone
out=Path('results/rhoptry_journal_review_2026_10_07_v2')
requests=[]
queries=[('preprint_history','https://api.biorxiv.org/details/biorxiv/10.1101/2025.10.16.682961'),('journal_bibliometrics','https://www.ebi.ac.uk/europepmc/webservices/rest/search?'+urllib.parse.urlencode({'query':'EXT_ID:42791346 AND SRC:MED','format':'json','resultType':'core'})),('preprint_bibliometrics','https://www.ebi.ac.uk/europepmc/webservices/rest/search?'+urllib.parse.urlencode({'query':'EXT_ID:41279910 AND SRC:MED','format':'json','resultType':'core'}))]
for name,url in queries:
    record={"name":name,"url":url,"retrieved_utc":datetime.now(timezone.utc).isoformat()}
    try:
        with urllib.request.urlopen(url,timeout=25) as response: payload=response.read(4<<20)
        parsed=json.loads(payload)
        (out/(name+".json")).write_bytes(payload)
        record.update(status="retrieved",sha256=hashlib.sha256(payload).hexdigest())
    except Exception as error:
        record.update(status="metadata_failed",error=f"{type(error).__name__}: {error}")
    requests.append(record)
(out/'lineage_requests.json').write_text(json.dumps(requests,indent=2))
requests
```
