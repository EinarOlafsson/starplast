# Ridge metric revision: unchanged native predictions and corrected baseline cards


Compare the initial diagnostic with canonical v2. Fix only constant-float variation detection; retain identical models, test entities, source truth, predictions and baseline numeric values. Distinct artifact identities must bind corrected metric code.

```python
import json, hashlib
from pathlib import Path
from starplast import artifacts as A
first=Path('results/ridge_value_pilot_2026_10_08')
second=Path('results/ridge_value_pilot_2026_10_08_v2')
read=lambda folder,name:json.loads((folder/'held_out'/name).read_text())
assert read(first,'rows.json')==read(second,'rows.json')
assert read(first,'model_state.json')==read(second,'model_state.json')
assert read(first,'baseline_state.json')==read(second,'baseline_state.json')
assert read(first,'own_kind_exclusions.json')==read(second,'own_kind_exclusions.json')
assert json.loads((first/'split.json').read_text())==json.loads((second/'split.json').read_text())
old=read(first,'card.json');new=read(second,'card.json')
assert old['metrics']==new['metrics']
before=old['extra']['baseline_comparison'];after=new['extra']['baseline_comparison']
assert before['mae_skill']==after['mae_skill'] and before['mse_skill']==after['mse_skill']
for partition in ('eligible_metrics','matched_metrics'):
    for metric,value in after[partition].items():
        if metric in ('top_decile_recall','bottom_decile_recall'):
            assert value is None
        else:
            assert value==before[partition][metric]
assert json.loads((first/'summary.json').read_text())['artifact_identity']!=json.loads((second/'summary.json').read_text())['artifact_identity']
print('All 1,101 predictions/truth/baseline values, ordered roles and fitted state exact; arbitrary constant-baseline decile metrics now unavailable, corrected code/artifact identities distinct.')
```
