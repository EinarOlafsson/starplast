# Split pilot integrity and declared regimes


Validate exact frozen inputs, output hashes, entity/group role separation and source exclusion selection. No biological accuracy claim.

```python
import ast,hashlib,json
from pathlib import Path
from starplast.splits import read_split
p=Path("results/split_protocols_2026_10_07_v2")
m=json.loads((p/"manifest.json").read_text())
for filename,digest in m["input_sha256"].items():
    assert hashlib.sha256(Path(filename).read_bytes()).hexdigest()==digest
for filename,digest in m["outputs"].items():
    assert hashlib.sha256((p/filename).read_bytes()).hexdigest()==digest
reports=json.loads((p/"pilot_report.json").read_text())
for report in reports:
    split,exclusions=read_split(p/(report["organism"]+"_"+report["target"]+".json"))
    assert sum(report["partitions"].values())==report["eligible_population"]
    assert len(split.assignments)==report["eligible_population"]
    split.guard_fit("feature_selection",exclusions.fit_entities)
for filename in ("starplast/splits.py","scripts/freeze_split_pilots.py"):
    ast.parse(Path(filename).read_text(),feature_version=(3,10))
print(m["summary"])
print("Inputs, outputs, cohorts, train-only source selection and Python 3.10 syntax verified")
```
