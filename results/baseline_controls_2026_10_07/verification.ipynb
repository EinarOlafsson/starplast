# Baseline/control fixture integrity and compatibility


Verify synthetic fixture identities and record software; no biological accuracy or capacity claim.

```python
import ast,hashlib,json,numpy,pandas
from pathlib import Path
p=Path("results/baseline_controls_2026_10_07")
m=json.loads((p/"manifest.json").read_text())
for filename,digest in m["input_sha256"].items():
    assert hashlib.sha256(Path(filename).read_bytes()).hexdigest()==digest
for filename,digest in m["outputs"].items():
    assert hashlib.sha256((p/filename).read_bytes()).hexdigest()==digest
for filename in ("starplast/baselines.py","scripts/verify_baseline_controls.py"):
    ast.parse(Path(filename).read_text(),feature_version=(3,10))
r=json.loads((p/"report.json").read_text())
assert r["labels"]["universe_sha256"]==r["values"]["universe_sha256"]
assert r["labels"]["evaluation_n"]==r["values"]["evaluation_n"]
assert r["positive_only"]["precision"] is None
print({"numpy":numpy.__version__,"pandas":pandas.__version__,"evaluation_n":r["labels"]["evaluation_n"],"truth_grade":m["truth_grade"],"mixing":r["network"]["mixing"]})
print("All inputs/outputs, matched denominators and Python 3.10 syntax verified")
```
