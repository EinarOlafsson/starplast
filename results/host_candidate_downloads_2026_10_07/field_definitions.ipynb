# RBC quantitative supplement field definitions


Read the published table header verbatim; whole RBC and white-ghost measurements stay distinct. Protein groups require ambiguity accounting rather than choosing their first accession.

```python
from pathlib import Path
import pandas as pd
path=Path('/media/carruthers/mnt3/claude/toxoplasma_projects/datasets/host/erythrocyte_proteome/PXD005200/publisher/pr7b00025_si_002.xlsx')
book=pd.ExcelFile(path)
for sheet in book.sheet_names:
    print(sheet)
    raw=pd.read_excel(book,sheet_name=sheet,header=None,nrows=3)
    print([(j,str(value)) for j,value in enumerate(raw.iloc[1]) if pd.notna(value)])
```
