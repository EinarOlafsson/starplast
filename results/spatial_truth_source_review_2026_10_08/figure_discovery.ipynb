# Primary spatial figure address and schema inspection


Inspect already verified XML and supplementary sources without assigning biological labels. Figure addresses are discovered, not guessed. The full paper remains in the external archive.

```python
import json, xml.etree.ElementTree as ET
from pathlib import Path
records=json.loads(Path('results/spatial_truth_source_review_2026_10_08/sources.json').read_text())
figures=[]
for record in records:
    article=ET.parse(record['fulltext']['path']).getroot()
    for figure in article.findall('.//fig'):
        label=figure.findtext('label','')
        hrefs=[e.get('{http://www.w3.org/1999/xlink}href') for e in figure.iter() if e.get('{http://www.w3.org/1999/xlink}href')]
        figures.append({'source':record['source_id'],'figure_id':figure.get('id'),'label':label,'hrefs':hrefs})
print(json.dumps(figures,indent=2))
```
