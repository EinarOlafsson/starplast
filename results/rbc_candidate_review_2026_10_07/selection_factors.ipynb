# RBC preference factors and biological comparison limits


Bibliometric factors are descriptive until biological admission. Copy counts, fraction PSMs and surface accessibility are different quantities. A missing common assay denominator remains unknown; no comprehensive claim follows from counting protein groups.

```python
import json,math
from pathlib import Path
from datetime import date
from starplast.dataset_selection import SelectionPolicy
source=Path('results/rbc_candidate_review_2026_10_07/review.json')
review=json.loads(source.read_text())
bib=review["bibliometrics"]
policy=SelectionPolicy()
age=(date.fromisoformat(bib["retrieved_utc"][:10])-date.fromisoformat(bib["first_publication_date"])).days/365.2425
factors={"policy":policy.to_dict(),"age_years":age,"citations_per_year":bib["citations_per_year"],"recency_factor":1+policy.recency_bonus*math.exp(-age/policy.recency_decay_years),"comprehensiveness_factor":None,"score":None,"status":"biological admission and common quantitative assay denominator unresolved","decision":review["decision"]}
Path('results/rbc_candidate_review_2026_10_07/selection_factors.json').write_text(json.dumps(factors,indent=2))
factors
```
