# Known-truth numeric error and matched baseline review


64.11 bounded metric prerequisite. Analytic known-truth controls verify constant baselines, small/constant cohorts, abstention and identical answered-row error skill. This is software/numerical validation, not an independent biological benchmark. Released source tables, fitted models and historical calibration records remain unchanged.

```python
import json, hashlib
import numpy as np, pandas as pd
from dataclasses import asdict
from starplast import scorecard as SC, record_scorecards as R, organisms as O
from pathlib import Path
out=Path('results/numeric_metric_review_2026_10_08')
scope=R.RecordScope(O.TOXOPLASMA,'trait_regression','known_numeric_control',SC.T_VALUES,'fixed known-truth control',17,'analytic_control','test','numeric-control-v1','synthetic_control','gene','All fixture truth explicitly known; no biology admitted',('Numerical fixture only',))
rows=pd.DataFrame({'entity':['g1','g2','g3'],'truth':[1.,3.,100.],'prediction':[1.,2.,np.nan],'baseline_prediction':[2.,2.,2.],'abstained':[False,False,True]})
card=R.aggregate(rows,scope,parameters={'quantity_unit':'fixture units','baseline_name':'fixed training-mean control'})
comparison=card['extra']['baseline_comparison']
assert comparison['mae_skill']==.5 and comparison['mse_skill']==.5
assert comparison['matched_metrics']['mae']==1.
assert comparison['eligible_metrics']['mae']==100/3
assert card['metrics']['rmse']==np.sqrt(.5) and card['counts']['answered']==2
constant=SC.values([2.5]*4,[1.,2.,3.,4.])
assert constant['mae']==1. and constant['rmse']==np.sqrt(1.25)
assert constant['r2']==0. and constant['nrmse']==1. and np.isnan(constant['spearman'])
single=SC.values([5.],[2.]); assert single['mae']==single['rmse']==3.
unknown=SC.values([np.nan]*2,[1.,2.]); assert unknown['value_coverage']==0. and np.isnan(unknown['rmse'])
summary={'fixture_card':card,'constant_baseline_metrics':R._finite(constant),'one_answer_metrics':R._finite(single),'no_answer_metrics':R._finite(unknown),'biological_admission':False}
(out/'summary.json').write_text(json.dumps(summary,indent=2,allow_nan=False)+'\n')
print(json.dumps(summary,indent=2,allow_nan=False))
```

Baseline error on the whole eligible cohort differs from error on the model’s answered cohort; skill uses only the latter. Zero-error baselines leave relative skill unavailable. Correlation remains unknown for constant predictions even while errors are defined. Row/parameter hashes pin the exact baseline and its name; training-only lineage is required separately at artifact construction.
