# Verify reviewed strategy capability snapshot


Check exact input/output identities and all native output meanings. This verifies a declaration/routing pilot, not biological performance.

```python
from pathlib import Path
import json, hashlib, ast
from starplast import capabilities as C
out = Path('results/capability_contracts_2026_10_08_v2')
m = json.loads((out / 'manifest.json').read_text())
sha = lambda p: hashlib.sha256(Path(p).read_bytes()).hexdigest()
for path, digest in m['input_sha256'].items():
    assert sha(path) == digest, path
for name, digest in m['outputs'].items():
    assert sha(out / name) == digest, name
native = json.loads((out / 'native_plans.json').read_text())
for plan in native:
    cap = C.get(plan['strategy'])
    assert plan['applicable'] and plan['validation_gaps']
    for output in plan['outputs']:
        cap.validate_output(C.Output(**output))
assert len(native) == 78
for path in ('starplast/capabilities.py', 'scripts/freeze_capability_catalog.py'):
    ast.parse(Path(path).read_text(), feature_version=(3, 10))
m['summary']
```
