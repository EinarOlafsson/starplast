# Inspect refused host withdrawal source association


The migration guard refused a protein cell that did not exactly match the ambiguous source gene score. No installed file was changed. Inspect the preserved source, installed table and original cache association; do not relax the exact-value guard or discard the source row.

```python
import pandas as pd, json
from pathlib import Path
base=Path('results/host_symbol_mapping_2026_10_08')
ambiguities=pd.read_parquet(base/'ambiguous_legacy_projection_review.parquet')
table=pd.read_parquet('starplast/data/hs_host_proteins.parquet').set_index('host_id')
cache=pd.read_csv('starplast/data/deposit_host_k562_rhoptry_screen.tsv',sep='\t',float_precision='round_trip')
fields=['rhoptry_discharge_score','rhoptry_discharge_beta','rhoptry_discharge_fdr']
rows=[]
for _,gene in ambiguities.iterrows():
    owner=gene.legacy_assigned_host_id
    cached=cache[cache.host_name.eq(gene.gene_symbol)]
    mismatch=[field for field in fields if table.at[owner,field]!=gene[field]]
    if mismatch:
        rows.append({'gene':gene.gene_symbol,'assumed_legacy_accession':owner,'actual_cache_owners':cached.host_id.tolist(),'different_fields':mismatch,'source':{field:float(gene[field]) for field in fields},'installed':{field:float(table.at[owner,field]) for field in fields},'cached':cached[fields].to_dict('records')})
print(json.dumps(rows,indent=2))
(base/'withdrawal_guard_diagnostic.json').write_text(json.dumps(rows,indent=2,allow_nan=False)+'\n')
```
